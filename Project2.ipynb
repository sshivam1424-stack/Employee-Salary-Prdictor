### Seprating the target variable(Annual Salary LPA) from the input features(X)


### and storing it in y for salary prediction

```python
X = data.drop('Annual_Salary_LPA',axis=1)
```

```python
y=data['Annual_Salary_LPA']
```

```python
X
```

```python
y
```

### Machine Learning start from here

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
X_test
```

```python
print(X_train.shape)
```

```python
print(X_test.shape)
```

```python
print(y_train.shape)
```

```python
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
```

```python
lr =LinearRegression()
lr.fit(X_train,y_train)

rf =RandomForestRegressor()
rf.fit(X_train,y_train)

xgb =GradientBoostingRegressor()
xgb.fit(X_train,y_train)

xg =XGBRegressor()
xg.fit(X_train,y_train)
```

```python
y_pred1 = lr.predict(X_test)
y_pred2 = rf.predict(X_test)
y_pred3 = xgb.predict(X_test)
y_pred4 = xg.predict(X_test)
```

```python
X_test.head()
```

```python
y_test.head()
```

```python
comparison = pd.DataFrame({
    'Actual Salary': y_test.values,
    'Linear Regression': y_pred1,
    'Random Forest': y_pred2,
    'Gradient Boosting': y_pred3,
    'XGBoost': y_pred4
})

comparison.head()
```

```python
y_test
```

```python
from sklearn import metrics
```

```python
score1 = metrics.r2_score(y_test,y_pred1)
score2 = metrics.r2_score(y_test,y_pred2)
score3 = metrics.r2_score(y_test,y_pred3)
score4 = metrics.r2_score(y_test,y_pred4)
```

```python
print(score1,score2,score3,score4)
```

```python
final_data = pd.DataFrame({'Models':['LR','RFR','GBR','XGB_R'],
                          "R2_SCORE":[score1,score2,score3,score4]})
```

```python
final_data
```

```python
sns.barplot(x=final_data['Models'],y=final_data['R2_SCORE'])
```
